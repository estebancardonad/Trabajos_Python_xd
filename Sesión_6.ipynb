#**Crear una Calculadora basica en la consola**



Intentar (try) pedir al usuario el primer número input.

Corregir (except) un ValueError si no escribe un número válido.

Pedir la operación (ej. "+", "-", "*", "/").

Intentar (try) pedir el segundo número input.

Capturar (except) un ValueError si no es válido.

Usar un bloque if/elif/else para comprobar qué operación eligió.

Intentar (try) realizar el cálculo.

Capturar (except) un ZeroDivisionError si el usuario intentó dividir por cero.

Mostrar el resultado.



```python
#Calculadorsiña Basica

print("Bienvenido a la Calculadora!")

while True:

    try:
        numero1 = float(input("Firmaaa, Ingresa el primer numero: "))
    except ValueError:
        print("Socio, un numero valido de bna")
        continue
    try:
        operacion = input("Ingresate la operacion ome maquinota (+,-,*,/): ")
    except ValueError:
        print("Niñooo de bna?, se equivoco de operacion jajaja, agarre seriedad")
        continue
    try:
        numero2 = float(input("Mostro, pone el Segundo Numero: "))
    except ValueError:
        print("Nea un numero pues")
        continue
    try:
        if operacion == "+":
          resultado = numero1 + numero2
        elif operacion == "-":
          resultado = numero1 - numero2
        elif operacion == "*":
          resultado = numero1 * numero2
        elif operacion == "/":
          resultado = numero1 / numero2
        print(f"El resultado es: {resultado}")
    except ZeroDivisionError:
        print("Nada de dividir por cero amigo mio")
    break
```
